#### Check Membership

```python
a = np.random.choice(data['Manufacturer'], 20, replace=True)
b = np.random.choice(data['Manufacturer'], 10, replace=True)
print(a)
print(b)
```

```python
# Find elements of a in b
a_in_b = np.isin(a, b)
a_in_b
```

```python
#  Union test
un = np.union1d(a, b)
un
```

```python
#  Intersection test
intsect = np.intersect1d(a, b)
intsect
```

```python
# Remove items of b from a
setdiff = np.setdiff1d(a, b)
setdiff
```

#### Code Challenge

```python
np.random.seed(100)
a2z=list('ABCDEFGHIJKLMNOPQRSTUVWXYZ')
a=np.random.choice(a2z, 20)
print(a)
b=np.random.choice(a2z, 20)
print(b)
c=np.random.choice(a2z, 20)
print(c)
```

```python
a_and_b = np.intersect1d(a, b)
print(a_and_b)
not_in_c = np.setdiff1d(a_and_b, c)
print(not_in_c)
```

```python
a_b = np.intersect1d(a, b)
a_b_c = np.intersect1d(a_b, c)
print(a_b_c)
```
