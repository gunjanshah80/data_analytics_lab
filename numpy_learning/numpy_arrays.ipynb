```python
import numpy as np
```

```python
li = [1, 2, 3, 4, 5]
arr = np.array(li)

print(arr.dtype)
print(arr * 2)
```

### **Reversing the array**

```python
arr[-1:-6:-1]
```

### **Two Dimensional Arrays**

```python
li2 = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
```

```python
arr2 = np.array(li2)
print(arr2.dtype)
print(arr2 * 3)
```

### **Array Indexing**

```python
arr2[0,:]
```

```python
arr2[:,1]
```

### **Convert datatype of an array**

```python
# converting from INT TO STR
arr2.astype('str')
```

```python
# converting from INT TO FLOAT
arr2.astype('float')
```

### **Mathematical Operations**

```python
# Multiply array by 2
arr3 = arr2 * 2
print("Multiply array by 2 gives:\n", arr3, "\n")

# Divide array by 4
arr4 =  arr2 / 4
print("Divide array by 4 gives:\n", arr4, "\n")

# Addition of two arrays of same dimensions
add_arr = arr2 + arr3
print("Addition of two arrays:\n", add_arr)
```

### **Convert back to list**

```python
new_list = arr2.tolist()
print(new_list)
```
