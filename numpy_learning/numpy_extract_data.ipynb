### **Extract Rows where any of the columns has missing value**

```python
# Any checks any of the columns, axis indicates rowwise check
data[np.isnan(data).any(axis=1),:]
```

```python
np.isnan(data).any(axis=1)
```

```python
# Drop all row where any of the columns has missing value present, using negate operator
data[~np.isnan(data).any(axis=1),:]
```

```python
# If a row contains a missing value, max return nan. Use nanmax instead to ignore such missing values
np.nanmax(data, axis=1)
```

```python
# If a column contains a missing value, max return nan. Use nanmax instead to ignore such missing values
np.nanmax(data, axis=0)
```

#### **If-Else logic using np.where() function**

```python
np.where(data[:,1]==1, data[:,4], data[:,4]/2)
```

```python
# Get the position of the maximum value in each row
np.argmax(data, axis=1)
```

```python
# Get the position of the maximum value in each row
max_pos=np.argmax(data, axis=1)
```

```python
[data[row,col] for row, col in enumerate(max_pos)]
```

#### **Get the position of the value where given condition is satisfied**

```python
pos = np.argwhere(data[:,1]==1)
pos[:10]
```
