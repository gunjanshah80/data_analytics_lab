### **Import CSV data file**

```python
data = np.genfromtxt("numpy_datasets/Mall_Customers.csv",
                     delimiter=",",
                     dtype="object",
                     skip_header=1)
data
```

```python
# Explicitly converting to float and do the opertions. Not a convenient way
output = data[:,3].astype('float') / data[:,2].astype('float')
print(output)
```

### **Better way to define datatype and import**

```python
dt = np.dtype({"names": ["CustomerID","Genre","Age","Annual_Income","Spending_Score"],
               "formats": [np.int16, 'U16', np.int16, np.int16, np.int16]
              })
```

```python
data = np.genfromtxt("numpy_datasets/Mall_Customers.csv",
                     delimiter=",",
                     dtype=dt,
                     skip_header=1)
data
```

```python
print("Total no. of dimension", data.ndim)
print(data[0]['Genre'])
print("Display all age values: \n", data['Age'])
```

## **Save output to a file**

```python
# Storing one array
np.save("TEMP/output.npy", data)
```

```python
#load data back from the file
data = np.load("TEMP/output.npy")
print(data[1]['Age'])
print(data['Genre'])
```

```python
# Storing multi dimensional array
np.savez("TEMP/outputs.npz", data, output)
```
