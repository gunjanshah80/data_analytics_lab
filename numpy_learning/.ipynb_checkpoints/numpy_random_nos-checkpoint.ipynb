```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### This is called Pseudo Random Numbers

```python
# Reproduce the same random numbers
np.random.seed(100)
np.random.rand(3)
```

```python
# Every execution generates new random numbers
np.random.random(size=(5,3)).round(2)
```

#### Random numers generation between range using Uniform Distribution

```python
dist_unif = np.random.uniform(1, 100, size=(10000))
dist_unif[:10].round(2)
```

```python
plt.hist(dist_unif)
```

```python
data = np.random.randint(1, 100, size=(10000))
data
```

```python
plt.hist(data)
```

#### Normal Distribution

```python
data_normal = np.random.normal(10, 2, size=(1000))
data_normal[:10].round(2)
```

```python
plt.hist(data_normal)
```

### Random Sampling

```python
arr = np.arange(100)
np.random.choice(arr, size=25, replace=False)
```

### Bootstrapping

```python
arr = np.arange(100)
boot = np.random.choice(arr, size=len(arr), replace=True)
boot
```

```python
np.unique(boot, return_counts=True)
```

### Binomial Distribution
#### n = no. of trials
#### p = probability of success

```python
result = np.random.binomial(n=10, p=0.5, size=1000000)
result[:10]
```

```python
plt.hist(result)
```

### Mini Challenge 1

```python
result = np.random.binomial(n=9, p=0.8, size=10000000)
result[:10]
```

```python
val = np.mean(result==6)
print(val)
```

### Mini Challenge 2

```python
data_nrml = np.random.normal(10, 3, size=(100))
data_nrml[:10]
```

```python
means = []
np.random.seed(100)
for i in range(100000):
    sample = np.random.choice(data_nrml, size=len(data_nrml), replace=True)
    means.append(np.mean(sample))

means[:10]
```

```python
stdev = np.std(means).round(2)
print(stdev)
```
